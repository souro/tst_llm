```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set(style="whitegrid")

def add_labels(ax, df, metric):
    for index, value in enumerate(df[metric]):
        ax.text(index, value + 0.01, f'{value:.2f}', ha='center', va='bottom', fontsize=9)

def plot_metrics(df, task, language, metrics, models, output_dir):
    num_metrics = len(metrics)
    fig, axes = plt.subplots(1, num_metrics, figsize=(6*num_metrics, 6), squeeze=False)

    df_filtered = df[df['Models'].isin(models)]

    for i, metric in enumerate(metrics):
        ax = axes[0, i]
        sns.barplot(x='Models', y=metric, data=df_filtered, ax=ax, palette="viridis")
        ax.set_title(f'{task} - {language} - {metric}')
        ax.set_xlabel('Models')
        ax.set_ylabel(metric)
        ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')

        add_labels(ax, df_filtered, metric)

    plt.tight_layout()
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    plt.savefig(os.path.join(output_dir, f'{task}_{language}.png'))
    plt.close()

def process_all_csvs(csv_files_info, output_dir='plots'):
    for info in csv_files_info:
        filepath = info['filepath']
        task = info['task']
        language = info['language']
        metrics = info['metrics']
        models = info['models']

        df = pd.read_csv(filepath)

        plot_metrics(df, task, language, metrics, models, output_dir)

csv_files_info = [
    # Sentiment Transfer - Automatic Metrics
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/auto_eval/sentiment_transfer_english.csv',
        'task': 'Automatic Evaluations - Sentiment Transfer',
        'language': 'English',
        'metrics': ['ACC', 'CS', 'BL'], #, 'AVG'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]

        # 'models': [
        #     'BLOOM-7B-ZS', 'BLOOMz-7B-ZS', 'ChatGLM2-6B-ZS', 'Falcon-7B-ZS', 'GPT-3.5-ZS',
        #     'Llama-7B-ZS', 'Llama-2-7B-ZS', 'Llama-2-Chat-7B-ZS', 'Llama-3-8B-ZS',
        #     'Llama-3-8B-Instruct-ZS', 'Mistral-7B-Instruct-ZS', 'OPT-6.7B-ZS',
        #     'Zephyr-7B-ZS', 'BLOOM-7B-FS', 'BLOOMz-7B-FS', 'ChatGLM2-6B-FS',
        #     'Falcon-7B-FS', 'GPT-3.5-FS', 'Llama-7B-FS', 'Llama-2-7B-FS',
        #     'Llama-2-Chat-7B-FS', 'Llama-3-8B-FS', 'Llama-3-8B-Instruct-FS',
        #     'Mistral-7B-Instruct-FS', 'OPT-6.7B-FS', 'BLOOM-7B-FT', 'BLOOMz-7B-FT',
        #     'ChatGLM2-6B-FT', 'Falcon-7B-FT', 'Llama-7B-FT', 'Llama-2-7B-FT',
        #     'Llama-2-13B-FT', 'Llama-3-8B-FT', 'OPT-6.7B-FT',
        #     'SOTA (Joint)', 'SOTA (Parallel)', 'SOTA (CLS-OP)', 'SOTA (KT)'
        # ]
    },
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/auto_eval/sentiment_transfer_hindi.csv',
        'task': 'Automatic Evaluations - Sentiment Transfer',
        'language': 'Hindi',
        'metrics': ['ACC', 'CS', 'BL'], #, 'AVG'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]

        # 'models': [
        #     'BLOOM-7B-ZS', 'BLOOMz-7B-ZS', 'ChatGLM2-6B-ZS', 'Falcon-7B-ZS', 'GPT-3.5-ZS',
        #     'Llama-7B-ZS', 'Llama-2-7B-ZS', 'Llama-2-Chat-7B-ZS', 'Llama-3-8B-ZS',
        #     'Llama-3-8B-Instruct-ZS', 'Mistral-7B-Instruct-ZS', 'OPT-6.7B-ZS',
        #     'Zephyr-7B-ZS', 'BLOOM-7B-FS', 'BLOOMz-7B-FS', 'ChatGLM2-6B-FS',
        #     'Falcon-7B-FS', 'GPT-3.5-FS', 'Llama-7B-FS', 'Llama-2-7B-FS',
        #     'Llama-2-Chat-7B-FS', 'Llama-3-8B-FS', 'Llama-3-8B-Instruct-FS',
        #     'Mistral-7B-Instruct-FS', 'OPT-6.7B-FS', 'BLOOM-7B-FT', 'BLOOMz-7B-FT',
        #     'ChatGLM2-6B-FT', 'Falcon-7B-FT', 'Llama-7B-FT', 'Llama-2-7B-FT',
        #     'Llama-2-13B-FT', 'Llama-3-8B-FT', 'OPT-6.7B-FT',
        #     'SOTA (Joint)', 'SOTA (Parallel)', 'SOTA (CLS-OP)', 'SOTA (KT)'
        # ]
    },
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/auto_eval/sentiment_transfer_bengali.csv',
        'task': 'Automatic Evaluations - Sentiment Transfer',
        'language': 'Bengali',
        'metrics': ['ACC', 'CS', 'BL'], #, 'AVG'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]

        # 'models': [
        #     'BLOOM-7B-ZS', 'BLOOMz-7B-ZS', 'ChatGLM2-6B-ZS', 'Falcon-7B-ZS', 'GPT-3.5-ZS',
        #     'Llama-7B-ZS', 'Llama-2-7B-ZS', 'Llama-2-Chat-7B-ZS', 'Llama-3-8B-ZS',
        #     'Llama-3-8B-Instruct-ZS', 'Mistral-7B-Instruct-ZS', 'OPT-6.7B-ZS',
        #     'Zephyr-7B-ZS', 'BLOOM-7B-FS', 'BLOOMz-7B-FS', 'ChatGLM2-6B-FS',
        #     'Falcon-7B-FS', 'GPT-3.5-FS', 'Llama-7B-FS', 'Llama-2-7B-FS',
        #     'Llama-2-Chat-7B-FS', 'Llama-3-8B-FS', 'Llama-3-8B-Instruct-FS',
        #     'Mistral-7B-Instruct-FS', 'OPT-6.7B-FS', 'BLOOM-7B-FT', 'BLOOMz-7B-FT',
        #     'ChatGLM2-6B-FT', 'Falcon-7B-FT', 'Llama-7B-FT', 'Llama-2-7B-FT',
        #     'Llama-2-13B-FT', 'Llama-3-8B-FT', 'OPT-6.7B-FT',
        #     'SOTA (Joint)', 'SOTA (Parallel)', 'SOTA (CLS-OP)', 'SOTA (KT)'
        # ]
    },
    # Detoxification - Automatic Metrics
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/auto_eval/detoxification_english.csv',
        'task': 'Automatic Evaluations - Detoxification',
        'language': 'English',
        'metrics': ['ACC', 'CS', 'BL'], #, 'AVG'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]

        # 'models': [
        #     'BLOOM-7B-ZS', 'BLOOMz-7B-ZS', 'ChatGLM2-6B-ZS', 'Falcon-7B-ZS', 'GPT-3.5-ZS',
        #     'Llama-7B-ZS', 'Llama-2-7B-ZS', 'Llama-2-Chat-7B-ZS', 'Llama-3-8B-ZS',
        #     'Llama-3-8B-Instruct-ZS', 'Mistral-7B-Instruct-ZS', 'OPT-6.7B-ZS',
        #     'Zephyr-7B-ZS', 'BLOOM-7B-FS', 'BLOOMz-7B-FS', 'ChatGLM2-6B-FS',
        #     'Falcon-7B-FS', 'GPT-3.5-FS', 'Llama-7B-FS', 'Llama-2-7B-FS',
        #     'Llama-2-Chat-7B-FS', 'Llama-3-8B-FS', 'Llama-3-8B-Instruct-FS',
        #     'Mistral-7B-Instruct-FS', 'OPT-6.7B-FS', 'BLOOM-7B-FT', 'BLOOMz-7B-FT',
        #     'ChatGLM2-6B-FT', 'Falcon-7B-FT', 'Llama-7B-FT', 'Llama-2-7B-FT',
        #     'Llama-2-13B-FT', 'Llama-3-8B-FT', 'OPT-6.7B-FT',
        #     'SOTA (Joint)', 'SOTA (Parallel)', 'SOTA (CLS-OP)', 'SOTA (KT)'
        # ]
    },
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/auto_eval/detoxification_hindi.csv',
        'task': 'Automatic Evaluations - Detoxification',
        'language': 'Hindi',
        'metrics': ['ACC', 'CS', 'BL'], #, 'AVG'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]

        # 'models': [
        #     'BLOOM-7B-ZS', 'BLOOMz-7B-ZS', 'ChatGLM2-6B-ZS', 'Falcon-7B-ZS', 'GPT-3.5-ZS',
        #     'Llama-7B-ZS', 'Llama-2-7B-ZS', 'Llama-2-Chat-7B-ZS', 'Llama-3-8B-ZS',
        #     'Llama-3-8B-Instruct-ZS', 'Mistral-7B-Instruct-ZS', 'OPT-6.7B-ZS',
        #     'Zephyr-7B-ZS', 'BLOOM-7B-FS', 'BLOOMz-7B-FS', 'ChatGLM2-6B-FS',
        #     'Falcon-7B-FS', 'GPT-3.5-FS', 'Llama-7B-FS', 'Llama-2-7B-FS',
        #     'Llama-2-Chat-7B-FS', 'Llama-3-8B-FS', 'Llama-3-8B-Instruct-FS',
        #     'Mistral-7B-Instruct-FS', 'OPT-6.7B-FS', 'BLOOM-7B-FT', 'BLOOMz-7B-FT',
        #     'ChatGLM2-6B-FT', 'Falcon-7B-FT', 'Llama-7B-FT', 'Llama-2-7B-FT',
        #     'Llama-2-13B-FT', 'Llama-3-8B-FT', 'OPT-6.7B-FT',
        #     'SOTA (Joint)', 'SOTA (Parallel)', 'SOTA (CLS-OP)', 'SOTA (KT)'
        # ]
    },
    # Sentiment Transfer - GPT-4 Based Evaluation
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/gpt4_eval/sentiment_transfer_english.csv',
        'task': 'GPT-4-based Evaluation - Sentiment Transfer',
        'language': 'English',
        'metrics': ['Sty.', 'Cont.', 'Flu.'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]
    },
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/gpt4_eval/sentiment_transfer_hindi.csv',
        'task': 'GPT-4-based Evaluation - Sentiment Transfer',
        'language': 'Hindi',
        'metrics': ['Sty.', 'Cont.', 'Flu.'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]
    },
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/gpt4_eval/sentiment_transfer_bengali.csv',
        'task': 'GPT-4-based Evaluation - Sentiment Transfer',
        'language': 'Bengali',
        'metrics': ['Sty.', 'Cont.', 'Flu.'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]
    },
    # Detoxification - GPT-4 Based Evaluation
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/gpt4_eval/detoxification_english.csv',
        'task': 'GPT-4-based Evaluation - Detoxification',
        'language': 'English',
        'metrics': ['Sty.', 'Cont.', 'Flu.'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]
    },
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/gpt4_eval/detoxification_hindi.csv',
        'task': 'GPT-4-based Evaluation - Detoxification',
        'language': 'Hindi',
        'metrics': ['Sty.', 'Cont.', 'Flu.'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'Mistral-7B-Instruct-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'Llama-3-8B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]
    },
    # Sentiment Transfer - Human Evaluation
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/human_eval/sentiment_transfer_english.csv',
        'task': 'Human Evaluation - Sentiment Transfer',
        'language': 'English',
        'metrics': ['Style', 'Content', 'Fluency'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]
    },
    {
        'filepath': '/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/human_eval/sentiment_transfer_hindi.csv',
        'task': 'Human Evaluation - Sentiment Transfer',
        'language': 'Hindi',
        'metrics': ['Style', 'Content', 'Fluency'],
        'models': [
            'GPT-3.5-ZS',
            'Llama-2-7B-Chat-ZS',
            'GPT-3.5-FS',
            'Mistral-7B-Instruct-FS',
            'Llama-2-13B-FT',
            'SOTA (Joint)',
            'SOTA (KT)'
        ]
    },
]

process_all_csvs(csv_files_info, output_dir='/content/drive/MyDrive/Colab Notebooks/tst_llms_results_plots/plots')
```
