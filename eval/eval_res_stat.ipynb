```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
import re
```

```python
human_eval_en = pd.read_csv('', index_col = False)
# human_eval_en = human_eval_en.loc[:, ~human_eval_en.columns.str.contains('^Unnamed')]

lang = 'en'
gpt4_eval_en = pd.read_csv(f'gpt4_eval_sentiment_{lang}.csv', index_col = False)

human_eval_hi = pd.read_csv('', index_col = False)
# human_eval_hi = human_eval_hi.loc[:, ~human_eval_hi.columns.str.contains('^Unnamed')]

lang = 'hi'
gpt4_eval_hi = pd.read_csv(f'gpt4_eval_sentiment_{lang}.csv', index_col = False)
```

```python
correlations_en = {
    'style_accuracy': pearsonr(human_eval_en['style_accuracy'], gpt4_eval_en['style_accuracy'])[0],
    'content_preservation': pearsonr(human_eval_en['content_preservation'], gpt4_eval_en['content_preservation'])[0],
    'fluency': pearsonr(human_eval_en['fluency'], gpt4_eval_en['fluency'])[0]
}

correlations_hi = {
    'style_accuracy': pearsonr(human_eval_hi['style_accuracy'], gpt4_eval_hi['style_accuracy'])[0],
    'content_preservation': pearsonr(human_eval_hi['content_preservation'], gpt4_eval_hi['content_preservation'])[0],
    'fluency': pearsonr(human_eval_hi['fluency'], gpt4_eval_hi['fluency'])[0]
}

print("Correlations (English):")
for key, value in correlations_en.items():
    print(f"{key}: {value:.2f}")

print("\nCorrelations (Hindi):")
for key, value in correlations_hi.items():
    print(f"{key}: {value:.2f}")

fig, axes = plt.subplots(2, 3, figsize=(15, 10))  # Adjusted figure size for two rows

sns.regplot(ax=axes[0, 0], x=human_eval_en['style_accuracy'], y=gpt4_eval_en['style_accuracy'])
axes[0, 0].set_title('Style Accuracy Correlation (EN)')
axes[0, 0].set_xlabel('human_eval_en Style Accuracy')
axes[0, 0].set_ylabel('gpt4_eval_en Style Accuracy')

sns.regplot(ax=axes[0, 1], x=human_eval_en['content_preservation'], y=gpt4_eval_en['content_preservation'])
axes[0, 1].set_title('Content Preservation Correlation (EN)')
axes[0, 1].set_xlabel('human_eval_en Content Preservation')
axes[0, 1].set_ylabel('gpt4_eval_en Content Preservation')

sns.regplot(ax=axes[0, 2], x=human_eval_en['fluency'], y=gpt4_eval_en['fluency'])
axes[0, 2].set_title('Fluency Correlation (EN)')
axes[0, 2].set_xlabel('human_eval_en Fluency')
axes[0, 2].set_ylabel('gpt4_eval_en Fluency')

# Hindi scatter plots
sns.regplot(ax=axes[1, 0], x=human_eval_hi['style_accuracy'], y=gpt4_eval_hi['style_accuracy'])
axes[1, 0].set_title('Style Accuracy Correlation (HI)')
axes[1, 0].set_xlabel('human_eval_hi Style Accuracy')
axes[1, 0].set_ylabel('gpt4_eval_hi Style Accuracy')

sns.regplot(ax=axes[1, 1], x=human_eval_hi['content_preservation'], y=gpt4_eval_hi['content_preservation'])
axes[1, 1].set_title('Content Preservation Correlation (HI)')
axes[1, 1].set_xlabel('human_eval_hi Content Preservation')
axes[1, 1].set_ylabel('gpt4_eval_hi Content Preservation')

sns.regplot(ax=axes[1, 2], x=human_eval_hi['fluency'], y=gpt4_eval_hi['fluency'])
axes[1, 2].set_title('Fluency Correlation (HI)')
axes[1, 2].set_xlabel('human_eval_hi Fluency')
axes[1, 2].set_ylabel('gpt4_eval_hi Fluency')

plt.tight_layout()
plt.savefig('figs/correlation_scatter_plots_en_hi.png', dpi=300)
# plt.show()

# combined_df = pd.DataFrame({
#     'df_eval_report_style_accuracy_en': human_eval_en['style_accuracy'],
#     'human_eval_en_style_accuracy_en': gpt4_eval_en['style_accuracy'],
#     'df_eval_report_content_preservation_en': human_eval_en['content_preservation'],
#     'human_eval_en_content_preservation_en': gpt4_eval_en['content_preservation'],
#     'df_eval_report_fluency_en': human_eval_en['fluency'],
#     'human_eval_en_fluency_en': gpt4_eval_en['fluency'],
#     'df_eval_report_style_accuracy_hi': human_eval_hi['style_accuracy'],
#     'human_eval_en_style_accuracy_hi': gpt4_eval_hi['style_accuracy'],
#     'df_eval_report_content_preservation_hi': human_eval_hi['content_preservation'],
#     'human_eval_en_content_preservation_hi': gpt4_eval_hi['content_preservation'],
#     'df_eval_report_fluency_hi': human_eval_hi['fluency'],
#     'human_eval_en_fluency_hi': gpt4_eval_hi['fluency']
# })

combined_df = pd.DataFrame({
    'human_eval_en_style_accuracy': human_eval_en['style_accuracy'],
    'gpt4_eval_en_style_accuracy': gpt4_eval_en['style_accuracy'],
    'human_eval_en_content_preservation': human_eval_en['content_preservation'],
    'gpt4_eval_en_content_preservation': gpt4_eval_en['content_preservation'],
    'human_eval_en_fluency': human_eval_en['fluency'],
    'gpt4_eval_en_fluency': gpt4_eval_en['fluency'],
    'human_eval_hi_style_accuracy': human_eval_hi['style_accuracy'],
    'gpt4_eval_hi_style_accuracy': gpt4_eval_hi['style_accuracy'],
    'human_eval_hi_content_preservation': human_eval_hi['content_preservation'],
    'gpt4_eval_hi_content_preservation': gpt4_eval_hi['content_preservation'],
    'human_eval_hi_fluency': human_eval_hi['fluency'],
    'gpt4_eval_hi_fluency': gpt4_eval_hi['fluency']
})

plt.figure(figsize=(12, 10))
corr_matrix = combined_df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Heatmap (English & Hindi)')
plt.tight_layout()
plt.savefig('figs/correlation_heatmap_en_hi.png', dpi=300)
# plt.show()
```

```python
auto_eval = pd.read_csv('auto_eval_sentiment.csv', index_col = False)
# auto_eval = pd.read_csv('auto_eval_toxicity.csv', index_col = False)
print(auto_eval.head())
```

```python
def classify_methodology(method):
    if re.search(r'zero_shot$', method):
        return 'zero_shot'
    elif re.search(r'few_shot$', method):
        return 'few_shot'
    elif re.search(r'finetune$', method):
        return 'finetune'
    elif re.search(r'sota$', method):
        return 'sota'
    else:
        return 'other'

auto_eval['Methodology Type'] = auto_eval['Methodology'].apply(classify_methodology)

numeric_columns = ["Sentiment Accuracy", "Similarity", "Bleu Score", "Avg"]
# numeric_columns = ["Style Accuracy", "Similarity", "Bleu Score", "Avg"]
grouping_columns = ['Methodology Type', 'Language']
df_grouped = auto_eval[grouping_columns + numeric_columns]

grouped_df = df_grouped.groupby(['Methodology Type', 'Language']).mean().reset_index()

fig, axs = plt.subplots(2, 2, figsize=(16, 12))

metrics = ["Sentiment Accuracy", "Similarity", "Bleu Score", "Avg"]
# metrics = ["Style Accuracy", "Similarity", "Bleu Score", "Avg"]
axes = axs.flatten()

for metric, ax in zip(metrics, axes):
    sns.barplot(data=grouped_df, x='Language', y=metric, hue='Methodology Type', ax=ax)
    ax.set_title(f'{metric} by Language and Methodology Type')
    ax.set_ylabel(metric)
    ax.set_xlabel('Language')

plt.tight_layout()
plt.savefig('figs/methodologies_comparison_sentiment.png', dpi=300)
plt.show()
```
