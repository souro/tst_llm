```python
import os
import pandas as pd
import random
```

```python
seed_value = 53
random.seed(seed_value)
```

```python
directory = '../output/sentiment/'

tasks = ['pos_to_neg', 'neg_to_pos']
methodologies = ['Llama2_7b_chat_zero_shot', 'gpt35_zero_shot',  'gpt35_few_shot', 'mistral_7b_instruct_few_shot', 'llama2_13b_finetune', 'joint']
# methodologies = ['mistral_7b_instruct_zero_shot', 'gpt35_zero_shot',  'gpt35_few_shot', 'mistral_7b_instruct_few_shot', 'llama3_8b_finetune', 'kt']
languages = ['bn'] #['en', 'hi', 'bn']
# languages = ['en', 'hi']

for lang in languages:
    selected_data = []

    for filename in os.listdir(directory):
        if filename.endswith('.csv'):            
            parts = filename.split('-')
            task = parts[0]
            current_lang = parts[1]
            methodology = parts[2].split('.')[0]  # Remove the '.csv' extension

            if task in tasks and methodology in methodologies and current_lang == lang:
            # if methodology in methodologies and current_lang == lang:
                
                df = pd.read_csv(os.path.join(directory, filename))

                # Randomly select 25 rows
                if task == 'pos_to_neg':
                    selected_rows = df.sample(25)
                elif task == 'neg_to_pos':
                    selected_rows = df.sample(25)

                # selected_rows = df.sample(50)

                selected_rows['methodology'] = methodology
                selected_rows['task'] = task

                selected_data.append(selected_rows)

    result_df = pd.concat(selected_data)

    result_df.to_csv(f'human_eval_sentiment_{lang}.csv', index=False)
```

```python
# For sentiment
def avg_score(methodology, metric_list):
    Llama2_7b_chat_zero_shot = list()
    gpt35_zero_shot = list()
    gpt35_few_shot = list()
    mistral_7b_instruct_few_shot = list()
    llama2_13b_finetune = list()
    joint = list()
    for idx, i in enumerate(methodology):
        if 'Llama2_7b_chat_zero_shot' in (i):
            Llama2_7b_chat_zero_shot.append(int(metric_list[idx]))
        elif 'gpt35_zero_shot' in (i):
            gpt35_zero_shot.append(int(metric_list[idx]))
        elif 'gpt35_few_shot' in (i):
            gpt35_few_shot.append(int(metric_list[idx]))
        elif 'mistral_7b_instruct_few_shot' in (i):
            mistral_7b_instruct_few_shot.append(int(metric_list[idx]))
        elif 'llama2_13b_finetune' in (i):
            llama2_13b_finetune.append(int(metric_list[idx]))
        elif 'joint' in (i):
            joint.append(int(metric_list[idx]))
    return sum(Llama2_7b_chat_zero_shot)/len(Llama2_7b_chat_zero_shot),sum(gpt35_zero_shot)/len(gpt35_zero_shot), sum(gpt35_few_shot)/len(gpt35_few_shot), sum(mistral_7b_instruct_few_shot)/len(mistral_7b_instruct_few_shot), sum(llama2_13b_finetune)/len(llama2_13b_finetune), sum(joint)/len(joint) 
```
